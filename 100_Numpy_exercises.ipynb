#### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆)

````python
n = 5
m = 4

a = np.ones((n, m))

for i in range(n):
    for j in range(m):
        if i != 0 and i != (n - 1) and j != (m - 1) and j != 0:
            a[i, j] = 0

a
````

#### 18. Create a 5x5 matrix with values 1,2,3,4 just below the diagonal (★☆☆)

#### 41. How to sum a small array faster than np.sum? (★★☆)

````python
%matplotlib inline
import timeit
import matplotlib.pyplot as plt

x = [10, 100, 1000]
t1 = []
t2 = []
t3 = []

for i in x:
    a = np.arange(i)
    print(f"On {i}-th turn")
    t1.append(timeit.timeit(lambda: np.sum(a), number = 1000000))
    t2.append(timeit.timeit(lambda: a.sum(), number = 1000000))
    t3.append(timeit.timeit(lambda: sum(a), number = 1000000))

for i in range(3):
    print(f"x = {x[i]}. [{t1[i]}, {t2[i]}, {t3[i]}]")

plt.plot(x, t1)
plt.plot(x, t2)
plt.plot(x, t3)
````

#### 48. Print the minimum and maximum representable values for each numpy scalar type (★★☆)

````python
numerical_types = [
    # Signed Integers
    np.int8, np.int16, np.int32, np.int64,
    # Unsigned Integers
    np.uint8, np.uint16, np.uint32, np.uint64,
    # Floats
    np.float16, np.float32, np.float64,
]

for code in numerical_types:
    dtype = np.dtype(code)
    info = np.iinfo(code) if np.issubdtype(dtype, np.integer) else np.finfo(code)
    print(f"{dtype}: min={info.min}, max={info.max}, bits={info.bits}")
````
